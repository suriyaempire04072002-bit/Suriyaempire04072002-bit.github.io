```python
df['date']=pd.to_datetime(df['date'])
df['week']=df['date'].dt.day_of_week
df
```

```python
df_store=df.groupby(['date','store_id']).agg({
    'amount':'sum',
    'quantity':'sum',
    'transaction_id':'count'
}).reset_index()
df_store.columns = ['Date', 'Store', 'Total_Sales', 'Total_Qty', 'Num_Transactions']
print(df_store)
```

```python
df.head()
df_highest=df.groupby('date')['amount'].sum().sort_values(ascending=False).head(1)
print(df_highest)
```
